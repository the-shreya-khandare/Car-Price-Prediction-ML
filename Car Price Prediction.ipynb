## **Library Imports**

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, mean_absolute_error
```

## **Data Loading**

```python
dataset = pd.read_csv("old_cardata.csv")
```

## **Data Preview**

```python
dataset.head()
```

```python
dataset.isnull().sum()
```

```python
dataset.info()
```

## **Categorical Variable Encoding using LabelEncoder**

## **Car_Name**

```python
Car_Name_le = LabelEncoder()
dataset["Car_Name"] = Car_Name_le.fit_transform(dataset["Car_Name"])
```

## **Fuel_Type**

```python
dataset["Fuel_Type"].unique()
```

```python
Fule_Type_le = LabelEncoder()
dataset["Fuel_Type"] = Fule_Type_le.fit_transform(dataset["Fuel_Type"])
```

## **Seller_Type**

```python
dataset["Seller_Type"].unique()
```

```python
Seller_Type_le = LabelEncoder()
dataset["Seller_Type"] = Seller_Type_le.fit_transform(dataset["Seller_Type"])
```

## **Transmission**

```python
dataset["Transmission"].unique()
```

```python
Transmission_le = LabelEncoder()
dataset["Transmission"] = Transmission_le.fit_transform(dataset["Transmission"])
```

## **Dataset After Label Encoding**

```python
dataset.head()
```

## **Input & Output Data**

```python
input_data = dataset.iloc[:,:-1]
output_data = dataset["Selling_Price"]
```

## **Standard Scaling of Input Features**

```python
ss = StandardScaler()
input_data = pd.DataFrame(ss.fit_transform(input_data), columns=input_data.columns)
```

## **Train-Test Split Configuration**

```python
X_train, X_test, y_train, y_test = train_test_split(input_data, output_data,test_size=0.2, random_state=42)
```

## **Random Forest Regressor Model Implementation**

```python
rf = RandomForestRegressor(n_estimators=100)
rf.fit(X_train, y_train)
rf.score(X_train, y_train)*100, rf.score(X_test, y_test)*100
```

## **Error Analysis of Random Forest Regression Model(MSE & MAE)**

```python
mean_squared_error(y_test, rf.predict(X_test)), mean_absolute_error(y_test, rf.predict(X_test))
```
